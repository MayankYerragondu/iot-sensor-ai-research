# Environmental Sensor Evaluation (Temperature, Humidity, Luminance)

This notebook evaluates LSTM-based forecasting models against baselines (ARIMA).

We measure:
- **RMSE** (Root Mean Squared Error)
- **MAE** (Mean Absolute Error)
- **PR-AUC** for anomaly detection alerts


import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np
import joblib
import boto3
import os

# Example: Load test dataset
df_test = pd.read_csv("../results/env_test.csv")
df_test.head()


# Actual vs predicted values (example format)
y_true = df_test[["temperature", "humidity", "lux"]].values
y_pred = df_test[["pred_temperature", "pred_humidity", "pred_lux"]].values

rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mae = mean_absolute_error(y_true, y_pred)

print(f"RMSE: {rmse:.3f}, MAE: {mae:.3f}")


# Plot true vs predicted for temperature
plt.figure(figsize=(10,5))
plt.plot(df_test["timestamp"], df_test["temperature"], label="True Temp")
plt.plot(df_test["timestamp"], df_test["pred_temperature"], label="Predicted Temp")
plt.legend()
plt.title("Temperature Forecasting - True vs Predicted")
plt.xticks(rotation=45)
plt.show()


## Summary

- RMSE/MAE values show forecasting accuracy.
- High error spikes may indicate anomalies.
- This evaluation helps confirm if multivariate LSTMs outperform ARIMA baselines.
